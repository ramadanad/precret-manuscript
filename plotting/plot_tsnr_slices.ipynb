# tSNR sub-figure in Figure 5 

```python
# libraries
import nibabel as nib
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path
```

```python
# load the tSNR maps for epi and bssfp
# path to the tSNR maps
# HARD-CODED PATH: fixed local PRF data root used by this example.
prf_dir = Path('/home/dramadan/data/prf_bids')
tsnr_dir = prf_dir / 'derivatives' / 'qametrics' / 'sub-01' / 'ses-03' / 'func' / 'tsnr'

tsnr_epi = nib.load(tsnr_dir / 'sub-01_ses-03_task-prf_run-01_acq-epi_desc-dummyRemoval-motionCorr_qa-tsnr.nii.gz').get_fdata()
tsnr_bssfp = nib.load(tsnr_dir / 'sub-01_ses-03_task-prf_run-01_acq-bssfp_desc-dummyRemoval-motionCorr_qa-tsnr.nii.gz').get_fdata()

output_dir = prf_dir / 'figures' / 'tsnr_slices'
output_dir.mkdir(parents=True, exist_ok=True)
```

```python
plt.figure(figsize=(10,5))
plt.imshow(tsnr_epi[:,:,30].T, cmap='viridis', origin='lower', vmin=0, vmax=40)
plt.title('tSNR map for EPI', fontsize=18)
cbar = plt.colorbar(label='tSNR', pad=0.04, ticks=np.arange(0, 41, 10), aspect=10) # less ticks for colorbar
cbar.ax.tick_params(labelsize=16, length=10) # increase tick label size and remove tick marks
cbar.set_label('tSNR', fontsize=18)
#  less ticks for colorbar
# remove x and y ticks
plt.xticks([])
plt.yticks([])
# remove frame
plt.gca().spines['top'].set_visible(False)
plt.gca().spines['right'].set_visible(False)
plt.gca().spines['bottom'].set_visible(False)
plt.gca().spines['left'].set_visible(False)
# remove white background 
plt.gca().set_facecolor('none')
# add the rectangle for the zoomed-in region
plt.gca().add_patch(plt.Rectangle((80, 5), 60, 45, edgecolor='magenta', facecolor='none', linewidth=3))
# save the figure without whitespace
plt.savefig(f'{output_dir}/tsnr_epi_sub-01_ses-03_run-01_ax30.svg', dpi=300, bbox_inches='tight', transparent=True)
```

```python
plt.figure(figsize=(6,5))
plt.imshow(tsnr_epi[80:140,5:50,30].T, cmap='viridis', origin='lower', vmin=0, vmax=40)
cbar.ax.tick_params(labelsize=16, length=10) # increase tick label size and remove tick marks
cbar.set_label('tSNR', fontsize=18)
#  less ticks for colorbar
# remove x and y ticks
plt.xticks([])
plt.yticks([])
# remove frame
plt.gca().spines['top'].set_visible(False)
plt.gca().spines['right'].set_visible(False)
plt.gca().spines['bottom'].set_visible(False)
plt.gca().spines['left'].set_visible(False)
# remove white background 
plt.gca().set_facecolor('none')
# save the figure without whitespace
plt.savefig(f'{output_dir}/tsnr_epi_sub-01_ses-03_run-01_ax30_zoom.svg', dpi=300, bbox_inches='tight', transparent=True)
```
